```python
import random

def print_board(state):
    n = len(state)
    for row in range(n):
        for col in range(n):
            if state[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()
    print()


def calculate_cost(state):
    n = len(state)
    conflicts = 0

    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def get_neighbors(state):
    n = len(state)
    neighbors = []

    for col in range(n):
        for row in range(n):
            if row != state[col]:
                new_state = state.copy()
                new_state[col] = row
                neighbors.append(new_state)

    return neighbors


def hill_climbing():

    n = 4
    current = [random.randint(0, n - 1) for _ in range(n)]

    print("Initial State:")
    print(current)
    print("Cost:", calculate_cost(current))
    print_board(current)

    while True:

        current_cost = calculate_cost(current)
        neighbors = get_neighbors(current)

        best_neighbor = min(neighbors, key=calculate_cost)
        best_cost = calculate_cost(best_neighbor)

        print("Current Cost:", current_cost)
        print("Best Neighbor:", best_neighbor)
        print("Neighbor Cost:", best_cost)
        print()

        if best_cost >= current_cost:
            if current_cost == 0:
                print("Goal configuration reached!")
                print_board(current)
                return current
            else:
                print("Hill Climbing stuck at local maximum/minimum.")
                print("Restarting...\n")
                current = [random.randint(0, n - 1) for _ in range(n)]
        else:
            current = best_neighbor


solution = hill_climbing()

print("Final Solution:")
print(solution)
print_board(solution)
```
