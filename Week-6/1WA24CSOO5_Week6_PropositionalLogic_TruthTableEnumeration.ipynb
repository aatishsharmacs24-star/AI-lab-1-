```python
from itertools import product

def biconditional(p, q):
    return p == q

def implication(p, q):
    return not p or q

def evaluate_kb(p, q, r):
    return biconditional(p, q) and (q or r) and implication(not p, r)

def tt_entails(kb, query, symbols, model):
    if len(model) == len(symbols):
        if kb(model):
            return query(model)
        return True

    symbol = symbols[len(model)]

    return (
        tt_entails(kb, query, symbols, model + {symbol: True})
        and
        tt_entails(kb, query, symbols, model + {symbol: False})
    )

symbols = ['P', 'Q', 'R']

print("P Q R | P<->Q | Q|R | ~P->R | KB | Q | P->Q | R->Q")

models = []

for p, q, r in product([True, False], repeat=3):
    a = biconditional(p, q)
    b = q or r
    c = implication(not p, r)
    kb = a and b and c

    q1 = q
    q2 = implication(p, q)
    q3 = implication(r, q)

    print(p, q, r, "|", a, "|", b, "|", c, "|", kb, "|", q1, "|", q2, "|", q3)

    if kb:
        models.append((p, q, r))

print("\nModels of KB:")
for model in models:
    print(model)

print("\nKB entails Q:", all(q for p, q, r in models))
print("KB entails P -> Q:", all(implication(p, q) for p, q, r in models))
print("KB entails R -> Q:", all(implication(r, q) for p, q, r in models))
```
