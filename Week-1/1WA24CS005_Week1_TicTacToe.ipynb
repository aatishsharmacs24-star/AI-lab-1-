```python
# Tic-Tac-Toe Game
# HUMAN = X
# COMPUTER = O

board = [' ' for _ in range(9)]


def display_board():
    print()
    print(board[0] + " | " + board[1] + " | " + board[2])
    print("--+---+--")
    print(board[3] + " | " + board[4] + " | " + board[5])
    print("--+---+--")
    print(board[6] + " | " + board[7] + " | " + board[8])
    print()


def check_winner(player):
    winning_positions = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        (0, 4, 8),
        (2, 4, 6)
    ]

    for a, b, c in winning_positions:
        if board[a] == board[b] == board[c] == player:
            return True

    return False


def computer_move():
    # Simple logic - no Minimax or other standard algorithm

    # First, try to win
    for i in range(9):
        if board[i] == ' ':
            board[i] = 'O'
            if check_winner('O'):
                return
            board[i] = ' '

    # Then, block the human
    for i in range(9):
        if board[i] == ' ':
            board[i] = 'X'
            if check_winner('X'):
                board[i] = 'O'
                return
            board[i] = ' '

    # Take the center
    if board[4] == ' ':
        board[4] = 'O'
        return

    # Otherwise take the first available position
    for i in range(9):
        if board[i] == ' ':
            board[i] = 'O'
            return


def tic_tac_toe():
    print("TIC-TAC-TOE")
    print("You are X")
    print("Computer is O")

    for i in range(9):
        display_board()

        # Human move
        while True:
            try:
                position = int(input("Enter position (1-9): "))

                if position < 1 or position > 9:
                    print("Enter a number between 1 and 9.")
                elif board[position - 1] != ' ':
                    print("Position already occupied.")
                else:
                    board[position - 1] = 'X'
                    break

            except ValueError:
                print("Please enter a valid number.")

        # Check human win
        if check_winner('X'):
            display_board()
            print("HUMAN WINS!")
            return

        # Check draw
        if ' ' not in board:
            display_board()
            print("GAME DRAW!")
            return

        # Computer move
        computer_move()

        # Check computer win
        if check_winner('O'):
            display_board()
            print("COMPUTER WINS!")
            return

        # Check draw
        if ' ' not in board:
            display_board()
            print("GAME DRAW!")
            return


tic_tac_toe()
```
