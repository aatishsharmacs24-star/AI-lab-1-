```python
# Vacuum Cleaner - Simple Reflex Agent

def simple_reflex_agent(location, room_status):

    if room_status[location] == "Dirty":
        return "SUCK"

    elif location == "A":
        return "MOVE RIGHT"

    else:
        return "MOVE LEFT"


def run_simple_reflex():
    room = {
        "A": "Dirty",
        "B": "Dirty"
    }

    location = "A"

    print("Initial State:")
    print(room)
    print("Vacuum Location:", location)

    while "Dirty" in room.values():

        action = simple_reflex_agent(location, room)

        print("\nLocation:", location)
        print("Action:", action)

        if action == "SUCK":
            room[location] = "Clean"

        elif action == "MOVE RIGHT":
            location = "B"

        elif action == "MOVE LEFT":
            location = "A"

    print("\nFinal State:")
    print(room)
    print("Both rooms are CLEAN!")


run_simple_reflex()
```
