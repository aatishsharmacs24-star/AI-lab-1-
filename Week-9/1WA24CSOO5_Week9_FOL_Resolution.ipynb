```python
def resolve(c1, c2):
    resolvents = set()

    for l1 in c1:
        for l2 in c2:
            if l1 == "!" + l2 or l2 == "!" + l1:
                new_clause = (c1 - {l1}) | (c2 - {l2})
                resolvents.add(frozenset(new_clause))

    return resolvents

def resolution(kb):
    clauses = set(kb)

    while True:
        new = set()

        clause_list = list(clauses)

        for i in range(len(clause_list)):
            for j in range(i + 1, len(clause_list)):
                resolvents = resolve(clause_list[i], clause_list[j])

                if frozenset() in resolvents:
                    return True

                new.update(resolvents)

        if new.issubset(clauses):
            return False

        clauses.update(new)

kb = {
    frozenset({"Alive(Anil)"}),
    frozenset({"!Alive(Anil)", "!Killed(Anil)"}),
    frozenset({"Eats(Anil,Peanuts)"}),
    frozenset({
        "!Eats(Anil,Peanuts)",
        "Killed(Anil)",
        "Food(Peanuts)"
    }),
    frozenset({
        "!Food(Peanuts)",
        "Likes(John,Peanuts)"
    }),
    frozenset({"!Likes(John,Peanuts)"})
}

if resolution(kb):
    print("Empty clause derived.")
    print("Query proved: John likes peanuts.")
else:
    print("Query could not be proved.")
```
