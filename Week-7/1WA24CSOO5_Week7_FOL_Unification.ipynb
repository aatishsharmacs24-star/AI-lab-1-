```python
def is_variable(x):
    return isinstance(x, str) and x in variables

variables = {"x", "y", "z", "u", "v", "w"}

def apply_substitution(term, subst):
    if is_variable(term):
        if term in subst:
            return apply_substitution(subst[term], subst)
        return term

    if isinstance(term, tuple):
        return tuple(apply_substitution(t, subst) for t in term)

    return term

def occurs(var, term, subst):
    term = apply_substitution(term, subst)

    if var == term:
        return True

    if isinstance(term, tuple):
        return any(occurs(var, t, subst) for t in term)

    return False

def unify(x, y, subst):
    x = apply_substitution(x, subst)
    y = apply_substitution(y, subst)

    if x == y:
        return subst

    if is_variable(x):
        if occurs(x, y, subst):
            return None
        subst[x] = y
        return subst

    if is_variable(y):
        if occurs(y, x, subst):
            return None
        subst[y] = x
        return subst

    if isinstance(x, tuple) and isinstance(y, tuple):
        if len(x) != len(y):
            return None

        for a, b in zip(x, y):
            subst = unify(a, b, subst)
            if subst is None:
                return None

        return subst

    return None

def show_result(a, b):
    result = unify(a, b, {})

    if result is None:
        print("Unification Failed")
    else:
        print("Unification Successful")
        print("Substitution:", result)
        print("Unified Expression:", apply_substitution(a, result))

show_result(("Eats", "x", "Apple"), ("Eats", "Riya", "y"))
show_result(("Knows", "John", "x"), ("Knows", "x", "Elisabeth"))
```
